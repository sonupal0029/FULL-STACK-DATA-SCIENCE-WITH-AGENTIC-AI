# LOGICAL OPERATOR

```python
x=20
y=10

print((x>=y) and (x==y))#1,0=0
print((x<=y) and (x!=y))#0,1=0
print((y<=x) and (y!=x))#1,1=1
print((x!=y) and (x<=y))#1,0=0
```

```python
x=100
y=300

print((x>=y) or (x==y))#0,0=0
print((x<=y) or (x!=y))#1,1=1
print((y<=x) or (y!=x))#0,1=1
print((x!=y) or (x>=y))#1,0=1

```
